```python
def message_encrypt(text, shift):
  encrypted_text = ""
  for char in text:
    if char.isalpha():
      shift = shift % 26
      if char.islower():
        encrypted_char = chr(ord('a') + (ord(char) - ord('a') + shift) % 26)
      else:
        encrypted_char = chr(ord('A') + (ord(char) - ord('A') + shift) % 26)
      encrypted_text += encrypted_char
    else:
      encrypted_text += char
  return encrypted_text
```

```python
# Insertion Sort

def insertion_sort(arr):
  for i in range(1,len(arr)):
    key = arr[i]
    j = i-1
    while j>=0 and arr[j]>key:
      arr[j+1] = arr[j]
      j -= 1
    arr[j+1] = key
  return arr

arr = [12,11,13,5,6]
print(insertion_sort(arr))
```

```python
# Selection Sort

def selection_sort(arr):
  for i in range(len(arr)):
    min_index = i
    for j in range(i+1,len(arr)):
      if arr[j]<arr[min_index]:
        min_index = j
        arr[i],arr[min_index] = arr[min_index],arr[i]
  return arr

arr = [12,11,13,5,6]
print(selection_sort(arr))
```

```python
# Bubble Sort

def bubble_sort(arr):
  for i in range(len(arr)):
    for j in range(len(arr)-i-1):
      if arr[j]>arr[j+1]:
        arr[j],arr[j+1] = arr[j+1],arr[j]
  return arr

arr = [12,11,13,5,6]
print(bubble_sort(arr))
```

```python
# Quick Sort

def quick_sort(arr):
  if len(arr) <= 1:
    return arr
  else:
    pivot = arr[len(arr)//2]
    left = [x for x in arr if x<pivot]
    middle = [x for x in arr if x==pivot]
    right = [x for x in arr if x>pivot]
    return quick_sort(left) + middle + quick_sort(right)

arr = [12,11,13,5,6]
print(quick_sort(arr))
```

```python
# Monty Hall Game Simulation

import random

# Function to simulate Monty Hall problem
def monty_hall_simulation(num_trials):
    stick_wins = 0
    switch_wins = 0

    for _ in range(num_trials):
        # Randomly place the car behind one of the three doors
        car_position = random.randint(0, 2)

        # Contestant makes an initial choice
        initial_choice = random.randint(0, 2)

        # Monty opens one of the remaining doors (that does not have the car)
        remaining_doors = [door for door in [0, 1, 2] if door != initial_choice and door != car_position]
        monty_opens = random.choice(remaining_doors)

        # The remaining unopened door
        remaining_door = [door for door in [0, 1, 2] if door != initial_choice and door != monty_opens][0]

        # Stick with the original choice
        if initial_choice == car_position:
            stick_wins += 1

        # Switch to the other door
        if remaining_door == car_position:
            switch_wins += 1

    return stick_wins, switch_wins

# Run the simulation for 10,000 trials
num_trials = 10000
stick_wins, switch_wins = monty_hall_simulation(num_trials)

stick_probability = stick_wins / num_trials
switch_probability = switch_wins / num_trials

stick_wins, switch_wins, stick_probability, switch_probability
```

```python
import random

# Function to simulate one round of the Gambler's Ruin problem
def gambler_ruin_simulation(initial_wealth, target_wealth, p, num_trials):
    reach_target_count = 0
    go_broke_count = 0

    for _ in range(num_trials):
        current_wealth = initial_wealth

        # Simulate until gambler either reaches the target or goes bankrupt
        while current_wealth > 0 and current_wealth < target_wealth:
            # Gambler wins with probability p, loses with probability 1 - p
            if random.random() < p:
                current_wealth += 1  # Win a round
            else:
                current_wealth -= 1  # Lose a round

        # Check final outcome
        if current_wealth >= target_wealth:
            reach_target_count += 1  # Reached the target
        else:
            go_broke_count += 1  # Went bankrupt

    # Calculate probabilities
    reach_target_prob = reach_target_count / num_trials
    go_broke_prob = go_broke_count / num_trials

    return reach_target_prob, go_broke_prob


# Parameters for the simulation
initial_wealth = 10       # Starting wealth
target_wealth = 20        # Target wealth
p = 0.485                  # Probability of winning a round (fair game)
num_trials = 10000        # Number of simulation trials

# Run the simulation
reach_target_prob, go_broke_prob = gambler_ruin_simulation(initial_wealth, target_wealth, p, num_trials)

# Print the results
print(f"Probability of reaching target wealth: {reach_target_prob:.4f}")
print(f"Probability of going broke: {go_broke_prob:.4f}")
```

```python
# Birthday Paradox

import math

# Function to calculate the probability of no shared birthday
def birthday_paradox(num_people):
    days_in_year = 365
    prob_no_shared_birthday = 1.0

    for i in range(num_people):
        prob_no_shared_birthday *= (days_in_year - i) / days_in_year

    return 1 - prob_no_shared_birthday


for j in range(1,30):
  print("when we have ",j," people the probability of shared birthday is ",birthday_paradox(j))
```

```python
import numpy as np
import pandas as pd

# Step 1: Create a dummy setup for a recommendation system dataset
# Here we create a user-item interaction matrix where rows represent users, and columns represent items.
# The values in the matrix represent ratings that users have given to items.

ratings_data = {
    'User1': [5, 3, 0, 1, 0],
    'User2': [4, 0, 0, 1, 1],
    'User3': [1, 1, 0, 5, 5],
    'User4': [1, 0, 4, 4, 4]
}

# Convert the data to a Pandas DataFrame for easy manipulation
ratings_df = pd.DataFrame(ratings_data, index=['Movie1', 'Movie2', 'Movie3', 'Movie4', 'Movie5']).T

# Convert the DataFrame to a NumPy array
ratings_matrix = ratings_df.values

# Step 2: Define the matrix factorization function using gradient descent
# We decompose the ratings matrix into two matrices, P and Q, such that P x Q approximates the original matrix

def matrix_factorization(R, P, Q, K, steps=5000, alpha=0.002, beta=0.02):
    # R: Original ratings matrix
    # P: User-feature matrix (initialized randomly)
    # Q: Item-feature matrix (initialized randomly)
    # K: Number of latent features
    # steps: Number of iterations to perform gradient descent
    # alpha: Learning rate
    # beta: Regularization parameter

    Q = Q.T  # Transpose Q to make the multiplication work

    for step in range(steps):
        for i in range(len(R)):
            for j in range(len(R[i])):
                if R[i][j] > 0:  # Only update for non-zero ratings
                    # Calculate the error for the current prediction
                    eij = abs(R[i][j] - np.dot(P[i, :], Q[:, j]))
                    # Update P and Q using gradient descent
                    for k in range(K):
                        P[i][k] += alpha * (2 * eij * Q[k][j] - beta * P[i][k])
                        Q[k][j] += alpha * (2 * eij * P[i][k] - beta * Q[k][j])
        # Calculate the total loss/error
        eR = np.dot(P, Q)
        e = 0
        for i in range(len(R)):
            for j in range(len(R[i])):
                if R[i][j] > 0:
                    e += (R[i][j] - np.dot(P[i, :], Q[:, j]))**2
                    for k in range(K):
                        e += (beta / 2) * (P[i][k]**2 + Q[k][j]**2)
        if e < 0.001:  # Stop if the total error is very small
            break
    return P, Q.T

# Step 3: Initialize the matrices P and Q with random values
num_users, num_items = ratings_matrix.shape
K = 2  # Number of latent features

P = np.random.rand(num_users, K)
Q = np.random.rand(num_items, K)

# Step 4: Perform matrix factorization using gradient descent
nP, nQ = matrix_factorization(ratings_matrix, P, Q, K)

# Step 5: Reconstruct the full matrix from the factors P and Q
predicted_ratings = np.dot(nP, nQ.T)  # Corrected the shape mismatch by transposing nQ

# Convert the result back to a DataFrame for better understanding
predicted_ratings_df = pd.DataFrame(predicted_ratings, index=ratings_df.index, columns=ratings_df.columns)

# Step 6: Print the original and predicted ratings matrices
print("Original Ratings Matrix:")
print(ratings_df)
print("\nPredicted Ratings Matrix:")
print(predicted_ratings_df)

# The predicted ratings matrix can be used to recommend items that a user hasn't rated yet
# by selecting the items with the highest predicted rating.
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
import itertools

# Load Data
headings = ['user_id', 'prod_id', 'rating', 'timestamp']
ratings = pd.read_csv('ratings_Electronics.csv',names = headings)
```

```python


# The given data has columns: ['user_id', 'prod_id', 'rating', 'timestamp']
# Drop timestamp as it is not needed for matrix factorization
ratings = ratings[['user_id', 'prod_id', 'rating']]
ratings = ratings.iloc[:10000]

# Min-Max Transformation
max_rating = ratings['rating'].max()
min_rating = ratings['rating'].min()
ratings['rating'] = (ratings['rating'] - min_rating) / (max_rating - min_rating)

# Create user and product ID mappings to indices
user_ids = ratings['user_id'].unique()
prod_ids = ratings['prod_id'].unique()

user_id_map = {user: idx for idx, user in enumerate(user_ids)}
prod_id_map = {prod: idx for idx, prod in enumerate(prod_ids)}

sub_user_id_map = dict(itertools.islice(user_id_map.items(), 500))
sub_prod_id_map = dict(itertools.islice(prod_id_map.items(), 100))

# Mapping user_id and prod_id in ratings to their respective indices
ratings['user_idx'] = ratings['user_id'].map(user_id_map)
ratings['prod_idx'] = ratings['prod_id'].map(prod_id_map)

# Create the User-Product Rating Matrix
num_users = len(user_ids)
num_prods = len(prod_ids)

R = np.zeros((num_users, num_prods))

for _, row in ratings.iterrows():
    R[row['user_idx'], row['prod_idx']] = row['rating']
```

```python
print(R.shape)
```

```python

# Define the number of latent features
latent_features = 10

# Initialize the User and Product matrices (U and V) randomly
np.random.seed(42)
U = np.random.normal(scale=1.0/latent_features, size=(num_users, latent_features))
V = np.random.normal(scale=1.0/latent_features, size=(num_prods, latent_features))

# Define hyperparameters
learning_rate = 0.01
regularization_param = 0.1
num_iterations = 100

# Store the cost for each iteration
costs = []

# Gradient Descent Optimization for Matrix Factorization
for iteration in range(num_iterations):
    # Calculate the predicted ratings
    R_pred = np.dot(U, V.T)

    # Calculate the error only for the known ratings
    error = (R - R_pred) * (R > 0)

    # Calculate cost (only for non-zero ratings)
    cost = np.sum((error[R > 0])**2) + regularization_param * (np.sum(U**2) + np.sum(V**2))
    costs.append(cost)

    # Compute the gradients
    U_grad = -2 * np.dot(error, V) + 2 * regularization_param * U
    V_grad = -2 * np.dot(error.T, U) + 2 * regularization_param * V

    # Update U and V using gradient descent
    U -= learning_rate * U_grad
    V -= learning_rate * V_grad

    # Print the cost every 100 iterations for tracking
    if (iteration + 1) % 10 == 0:
        print(f'Iteration {iteration + 1}/{num_iterations}, Cost: {cost}')

# Plotting the Cost Function
plt.plot(costs)
plt.xlabel('Iteration')
plt.ylabel('Cost')
plt.title('Training Cost Over Iterations')
plt.show()

# Evaluate the model
train_rmse = np.sqrt(mean_squared_error(R[R > 0], R_pred[R > 0]))
print(f'Training RMSE: {train_rmse}')

# Recommendation Function
def recommend_products(user_idx, top_n=5):
    """
    Recommend top_n products for a given user based on the predicted ratings.
    """
    user_ratings = R_pred[user_idx]
    already_rated = R[user_idx] > 0

    recommendations = [(prod_id, user_ratings[idx]) for idx, prod_id in enumerate(prod_ids) if not already_rated[idx]]
    recommendations = sorted(recommendations, key=lambda x: x[1], reverse=True)[:top_n]
    return recommendations

# Get Recommendations for a Sample User
sample_user_idx = 0
recommended_products = recommend_products(sample_user_idx, top_n=3)
print(f'Recommended Products for User {user_ids[sample_user_idx]}: {recommended_products}')

# Explanation:
# 1. We load the CSV file and extract necessary columns, dropping timestamp as it's not needed.
# 2. We map user and product IDs to numerical indices.
# 3. We initialize U and V matrices randomly for 2 latent features.
# 4. We use gradient descent to iteratively optimize U and V.
# 5. The cost function includes a regularization term to avoid overfitting.
# 6. We calculate predicted ratings and recommend top N products for any given user.

# Note:
# The gradient descent implementation is quite simple here, and the cost function may take some time to converge.
#In practice, there are more efficient ways to perform matrix factorization, such as using Alternating Least Squares (ALS).
```

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures,StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

# Step 1: Generate Non-Linear Synthetic Data (2nd Degree Polynomial Data)
np.random.seed(42)
X = np.linspace(-10, 10, 5000)  # 100 data points between -5 and 5
Y = 2 * X**2 - 3 * X + 5   # 2nd degree polynomial relationship with high noise

# Reshape X for model compatibility
X = X.reshape(-1, 1)

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

# Step 2: Create a Polynomial Regression Model of Degree 25 Without Regularization (Overfitting Scenario)
poly_degree = 25
poly_reg_model = make_pipeline(PolynomialFeatures(degree=poly_degree), LinearRegression())
poly_reg_model.fit(X_train, y_train)
y_pred_poly = poly_reg_model.predict(X_test)

# Calculate MSE for Unregularized Polynomial Regression Model
mse_poly = mean_squared_error(y_test, y_pred_poly)
print(f'Mean Squared Error (Unregularized Polynomial Regression - Degree 25): {mse_poly:.4f}')

# Step 3: Create a Polynomial Regression Model of Degree 25 With Ridge (L2) Regularization
ridge_reg_model = make_pipeline(PolynomialFeatures(degree=poly_degree),StandardScaler(), Ridge(alpha=10.0))
ridge_reg_model.fit(X_train, y_train)
y_pred_ridge = ridge_reg_model.predict(X_test)

# Calculate MSE for Ridge Regularized Polynomial Regression Model
mse_ridge = mean_squared_error(y_test, y_pred_ridge)
print(f'Mean Squared Error (Ridge Regularized Polynomial Regression - Degree 25): {mse_ridge:.4f}')

# Step 4: Create a Polynomial Regression Model of Degree 25 With Lasso (L1) Regularization
lasso_reg_model = make_pipeline(PolynomialFeatures(degree=poly_degree),StandardScaler(), Lasso(alpha=0.5, max_iter=10000))
lasso_reg_model.fit(X_train, y_train)
y_pred_lasso = lasso_reg_model.predict(X_test)

# Calculate MSE for Lasso Regularized Polynomial Regression Model
mse_lasso = mean_squared_error(y_test, y_pred_lasso)
print(f'Mean Squared Error (Lasso Regularized Polynomial Regression - Degree 25): {mse_lasso:.4f}')

# Step 5: Plot the Results with Scatter Plots for Better Visualization
plt.figure(figsize=(8, 6))
plt.scatter(X, Y, color='blue', alpha=0.5, label='Original Data')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Original Data')
plt.legend()
plt.show()

# Scatter plot for unregularized polynomial regression results
plt.figure(figsize=(8, 6))
plt.scatter(X_test, y_pred_poly, color='red', alpha=0.6, label='Unregularized Polynomial Regression (Degree 25)')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Unregularized Polynomial Regression Data')
plt.legend()
plt.show()

# Scatter plot for Ridge regression results
plt.figure(figsize=(8, 6))
plt.scatter(X_test, y_pred_ridge, color='green', alpha=0.6, label='Ridge Regularized Polynomial Regression (Degree 15)')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Ridge Regularized Polynomial Regression')
plt.legend()
plt.show()

# Scatter plot for Lasso regression results
plt.figure(figsize=(8, 6))
plt.scatter(X_test, y_pred_lasso, color='orange', alpha=0.6, label='Lasso Regularized Polynomial Regression (Degree 15)')

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Lasso Regularized Polynomial Regression (Degree 15)')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Generate Points
def generate_points(num_points):
    np.random.seed(42)  # Set a seed for reproducibility
    # generate only integer numbers

    x = np.random.randint(-10, 11, size=num_points)
    y = np.random.randint(-10, 11, size=num_points)
    return x, y

# Step 2: Solve for Polynomial Coefficients
def fit_polynomial(x, y):
    n = len(x)
    A = np.vander(x, n, increasing=True)  # Vandermonde matrix
    coeffs = np.linalg.solve(A, y)       # Solve linear system
    return coeffs

# Step 3: Evaluate Polynomial
def evaluate_polynomial(coeffs, x):
    return np.polyval(coeffs[::-1], x)  # Evaluate polynomial using coefficients

# Step 4: Iteratively Fit and Plot
def iterative_polynomial_fitting(x, y):
    for i in range(2, len(x) + 1):  # Start with 2 points up to all points
        print(f"\nFitting Polynomial with {i} points:")
        x_subset = x[:i]
        y_subset = y[:i]
        coeffs = fit_polynomial(x_subset, y_subset)
        print(f"Coefficients: {coeffs[::-1]}")  # Reverse for traditional polynomial order

        # Plot the current polynomial
        plot_polynomial(x, y, x_subset, y_subset, coeffs, i)

# Step 5: Plot
def plot_polynomial(x_all, y_all, x_subset, y_subset, coeffs, num_points):
    x_range = np.linspace(min(x_all) - 1, max(x_all) + 1, 500)
    y_range = evaluate_polynomial(coeffs, x_range)

    plt.scatter(x_all, y_all, color='gray', alpha=0.5, label='All Points')
    plt.scatter(x_subset, y_subset, color='red', label=f'Selected {num_points} Points')
    plt.plot(x_range, y_range, color='blue', label=f'Fitted Polynomial ({num_points-1} degree)')
    plt.title(f'Polynomial Fit with {num_points} Points')
    plt.legend()
    plt.grid(True)
    plt.show()

# Main
num_points = 10  # Total number of points
# generate the points only with integer values

x, y = generate_points(num_points)
print(f"Original Points (x, y): {list(zip(x, y))}")
iterative_polynomial_fitting(x, y)
```

```python
# Define a random 10x10 matrix
matrix_10x10 = np.random.rand(10, 10)

# Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(matrix_10x10)
max_eigenvalue_index = np.argmax(np.abs(eigenvalues))
dominant_eigenvector = eigenvectors[:, max_eigenvalue_index]

# Normalize the dominant eigenvector
dominant_eigenvector = dominant_eigenvector / np.linalg.norm(dominant_eigenvector)

# Define a random 10D vector
vector_10d = np.random.rand(10)
vector_10d = normalize(vector_10d)

# Track the convergence
differences = []

# Iteratively apply the matrix and normalize, compute differences
for _ in range(50):  # 50 iterations for better convergence observation
    vector_10d = normalize(np.dot(matrix_10x10, vector_10d))
    difference = np.linalg.norm(vector_10d - dominant_eigenvector)
    differences.append(difference)

# Plot the convergence
plt.figure(figsize=(10, 6))
plt.plot(differences, marker='o')
plt.title("Convergence of Vector to Dominant Eigenvector")
plt.xlabel("Iteration")
plt.ylabel("Difference from Dominant Eigenvector")
plt.grid()
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# Create a simple graph for demonstration
G = nx.Graph()
edges = [(1, 2), (2, 3), (3, 4), (4, 1), (1, 3), (2, 4)]  # A square with diagonals
G.add_edges_from(edges)

# Function to highlight a walk on the graph
def highlight_walk(graph, walk_edges, title):
    pos = nx.spring_layout(graph)  # Position for nodes
    plt.figure(figsize=(6, 6))
    nx.draw(graph, pos, with_labels=True, node_color="lightblue", node_size=700, edge_color="gray")
    nx.draw_networkx_edges(graph, pos, edgelist=walk_edges, edge_color="red", width=2)
    plt.title(title)
    plt.show()

# Define different types of walks
walks = {
    "Simple Walk": [(1, 2), (2, 3), (3, 4)],  # A walk without revisiting nodes or edges
    "Trail": [(1, 2), (2, 3), (3, 1)],  # A walk with no repeated edges, nodes can repeat
    "Path": [(1, 2), (2, 3), (3, 4)],  # A walk with no repeated nodes or edges
    "Cycle": [(1, 2), (2, 3), (3, 1)],  # A closed walk with no repeated nodes or edges (except start and end)
    "Circuit": [(1, 2), (2, 4), (4, 1), (1, 3), (3, 2)],  # A closed trail where edges don't repeat
}

# Illustrate each walk
for walk_type, edges in walks.items():
    highlight_walk(G, edges, walk_type)
```

```python
import networkx as nx
import random

def random_walk(graph, start_node, steps):
    """
    Perform a random walk on a graph.

    Parameters:
    - graph: The input graph (NetworkX Graph or DiGraph)
    - start_node: The node where the walk starts
    - steps: Number of steps to take

    Returns:
    - walk: List of nodes visited during the random walk
    """
    walk = [start_node]  # Start the walk at the given node
    for _ in range(steps - 1):  # Perform steps-1 steps
        current_node = walk[-1]
        neighbors = list(graph.neighbors(current_node))  # Get neighbors of the current node
        if not neighbors:
            break  # If no neighbors, stop the walk
        next_node = random.choice(neighbors)  # Randomly select the next node
        walk.append(next_node)
    return walk

# Example Usage
if __name__ == "__main__":
    # Create a simple graph
    G = nx.Graph()
    edges = [(1, 2), (2, 3), (3, 4), (4, 1), (1, 3), (2, 4)]  # A square with diagonals
    G.add_edges_from(edges)

    # Perform a random walk
    start_node = 1
    steps = 10
    walk = random_walk(G, start_node, steps)

    print(f"Random Walk (Start Node: {start_node}, Steps: {steps}):")
    print(walk)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Generate dummy data
np.random.seed(42)
study_hours = np.random.uniform(0, 10, 100)  # Study hours between 0 and 10
# Pass/fail labels based on a noisy threshold
def pass_fail(hours):
    if hours < 2:
        return 0  # Fail
    elif hours > 8:
        return 1  # Pass
    else:
        return np.random.choice([0, 1], p=[0.4, 0.6])  # Uncertain region

labels = np.array([pass_fail(h) for h in study_hours])

# Visualize the data
plt.figure(figsize=(10, 6))
plt.scatter(study_hours, labels, c=labels, cmap="bwr", alpha=0.7, edgecolor="k")
plt.axvline(x=2, color="gray", linestyle="--", label="Fail Threshold")
plt.axvline(x=8, color="gray", linestyle="--", label="Pass Threshold")
plt.title("Student Outcomes Based on Study Hours")
plt.xlabel("Study Hours")
plt.ylabel("Pass (1) / Fail (0)")
plt.legend()
plt.show()

# Step function model
def step_function(hours):
    return np.where(hours > 5, 1, 0)  # Hard decision boundary at 5 hours

# Predictions using step function
step_predictions = step_function(study_hours)

# Logistic regression model (sigmoid function)
log_reg = LogisticRegression()
study_hours_reshaped = study_hours.reshape(-1, 1)
log_reg.fit(study_hours_reshaped, labels)
sigmoid_predictions = log_reg.predict(study_hours_reshaped)

# Plot the decision boundaries
x_range = np.linspace(0, 10, 300)
sigmoid_boundary = log_reg.predict_proba(x_range.reshape(-1, 1))[:, 1]  # Probability of passing

plt.figure(figsize=(10, 6))
plt.scatter(study_hours, labels, c=labels, cmap="bwr", alpha=0.7, edgecolor="k", label="Data")
plt.plot(x_range, step_function(x_range), label="Step Function Boundary", color="orange", linestyle="--")
plt.plot(x_range, sigmoid_boundary, label="Sigmoid Function Boundary", color="green")
plt.title("Decision Boundaries: Step vs Sigmoid")
plt.xlabel("Study Hours")
plt.ylabel("Pass Probability")
plt.legend()
plt.show()

# Accuracy comparison
step_accuracy = accuracy_score(labels, step_predictions)
sigmoid_accuracy = accuracy_score(labels, sigmoid_predictions)
print(f"Step Function Accuracy: {step_accuracy:.2f}")
print(f"Sigmoid Function Accuracy: {sigmoid_accuracy:.2f}")
```

```python
import torch.nn as nn

class MNIST_MLP(nn.Module):
  def __init__(self):
    super().__init__()
    self.flatten = nn.Flatten() # convert the matrix to a vector
    self.linear_relu_stack = nn.Sequential(
        nn.Linear(28*28,512),
        nn.ReLU(),
        nn.Linear(512,512),
        nn.ReLU(),
        nn.Linear(512,10)
    )

  def forward(self,x):
    x = self.flatten(x)
    logits = self.linear_relu_stack(x)
    return logits
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# Define the LeNet architecture
class LeNet(nn.Module):
    def __init__(self):
        super(LeNet, self).__init__()
        self.conv1 = nn.Conv2d(1, 6, kernel_size=5, stride=1, padding=2)  # Input: 1x28x28, Output: 6x28x28
        self.conv2 = nn.Conv2d(6, 16, kernel_size=5, stride=1)  # Input: 6x28x28, Output: 16x24x24
        self.fc1 = nn.Linear(16 * 5 * 5, 120)  # Fully connected layer
        self.fc2 = nn.Linear(120, 84)
        self.fc3 = nn.Linear(84, 10)  # Output layer for 10 classes

    def forward(self, x):
        x = torch.relu(self.conv1(x))
        x = torch.max_pool2d(x, kernel_size=2, stride=2)  # Output: 6x14x14
        x = torch.relu(self.conv2(x))
        x = torch.max_pool2d(x, kernel_size=2, stride=2)  # Output: 16x5x5
        x = x.view(-1, 16 * 5 * 5)  # Flatten
        x = torch.relu(self.fc1(x))
        x = torch.relu(self.fc2(x))
        x = self.fc3(x)  # Output logits
        return x

# Data preparation
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))  # Normalize to mean 0, std 1
])

train_dataset = datasets.MNIST(root="./data", train=True, download=True, transform=transform)
test_dataset = datasets.MNIST(root="./data", train=False, download=True, transform=transform)

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=1000, shuffle=False)

# Model, loss function, optimizer
model = LeNet()
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters(), lr=0.01, momentum=0.9)

# Training loop
def train(model, device, train_loader, optimizer, criterion, epoch):
    model.train()
    for batch_idx, (data, target) in enumerate(train_loader):
        data, target = data.to(device), target.to(device)

        # Forward pass
        output = model(data)
        loss = criterion(output, target)

        # Backward pass
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        if batch_idx % 100 == 0:
            print(f"Train Epoch: {epoch} [{batch_idx * len(data)}/{len(train_loader.dataset)} ({100. * batch_idx / len(train_loader):.0f}%)]\tLoss: {loss.item():.6f}")

# Testing loop
def test(model, device, test_loader, criterion):
    model.eval()
    test_loss = 0
    correct = 0
    with torch.no_grad():
        for data, target in test_loader:
            data, target = data.to(device), target.to(device)
            output = model(data)

            # Sum up batch loss
            test_loss += criterion(output, target).item()

            # Get the index of the max log-probability
            pred = output.argmax(dim=1, keepdim=True)
            correct += pred.eq(target.view_as(pred)).sum().item()

    test_loss /= len(test_loader.dataset)
    print(f"\nTest set: Average loss: {test_loss:.4f}, Accuracy: {correct}/{len(test_loader.dataset)} ({100. * correct / len(test_loader.dataset):.0f}%)\n")

# Main training/testing process
num_epochs = 10
for epoch in range(1, num_epochs + 1):
    train(model, device, train_loader, optimizer, criterion, epoch)
    test(model, device, test_loader, criterion)
```
